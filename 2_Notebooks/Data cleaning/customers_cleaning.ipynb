#### Cols to filled

```python
nv_df
```

```python
cols =['segment','region','join_date']

for i in cols:
    df[i] = df[i].fillna(df[i].mode()[0])
    
```

```python
df['email'] = df['email'].fillna(
    'customer' + df['customer_id'].str.extract(r'(\d+)')[0].astype(int).astype(str) + '@example.com'
)
```
